```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import statsmodels.api as sm
from sklearn.linear_model import LinearRegression
```

```python
data = pd.read_csv("../data/1.02. Multiple linear regression.csv")
```

```python
data
```

```python
data.describe()
```

```python
# 変数を定義する
y = data["GPA"]
x = data[["SAT", "Rand 1,2,3"]]
```

```python
reg = LinearRegression()
reg.fit(x, y)
```

```python
# 係数, 切片, 決定係数を表示する
(reg.coef_), (reg.intercept_), (reg.score(x, y))
```

```python
# 決定係数を計算する
reg.score(x, y)
```

```python
from sklearn.feature_selection import f_regression
```

```python
f_regression(x, y)
```

```python
reg_summary = pd.DataFrame(data=x.columns.values, columns=["Features"])
reg_summary
```

```python
# まとめ表を作成する
reg_summary["Coefficients"] = reg.coef_
reg_summary["p-values"] = f_regression(x, y)[1]
```

```python
reg_summary
```

```python
# 標準化
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler.fit(x)
```

```python
x_scaled = scaler.transform(x)
x_scaled
```

```python
# 標準化したデータで回帰分析を行う
reg = LinearRegression()
reg.fit(x_scaled, y)
```

```python
# 係数, 切片, 決定係数を表示する
reg.coef_, reg.intercept_, reg.score(x_scaled, y)
```

```python
reg_summary = pd.DataFrame([["Bias"], ["SAT"], ["Rand 1,2,3"]], columns=["Features"])
reg_summary["Weights"] = reg.intercept_, reg.coef_[0], reg.coef_[1]
reg_summary

# 重みが大きいのはSATの方である
# 消してもモデルに影響はない、この重みがあることにより、P値を表示するメソッドがない
```

```python
# 予測
new_data = pd.DataFrame(data=[[1700, 2], [1800, 1]], columns=["SAT", "Rand 1,2,3"])
new_data_scaled = scaler.transform(new_data)
new_data_scaled
```

```python
reg.predict(new_data_scaled)
```
